```python
import random
from datetime import datetime, timedelta
```

```python
from faker import Faker
import faker_commerce

fake_client = Faker('ru_RU')

fake_product = Faker()

fake_product.add_provider(faker_commerce.Provider)
```

```python
NUM_OF_CLIENTS = 100
NUM_OF_CATEGORIES = 20
NUM_OF_PRODUCTS = 100
```

```python
possible_loyalty_statuses = ['VIP','Brilliant', 'Platimun', 'Gold', 'Silver']
```

```python
# генерируем данные о клиентах
for i in range(NUM_OF_CLIENTS):
    print(f"INSERT INTO users(first_name,last_name,email,phone,registration_date,loyalty_status) values(\
    '{fake_client.first_name()}', \
    '{fake_client.last_name()}', \
    '{fake_client.email()}', \
    '{fake_client.phone_number()}', \
    '{fake_client.date_between(start_date='-3y', end_date='now')}', \
    '{random.choice(possible_loyalty_statuses)}' \
     );"
         )
```

```python
# генерируем данные о продуктовых категориях
```

```python
for i in range(NUM_OF_CATEGORIES):
    print(f"INSERT INTO productcategories(name) values('{fake_product.unique.ecommerce_category()}');"
         )
```

```python
# генерируем данные о продуктах
```

```python
for i in range(NUM_OF_PRODUCTS):
    print(f"INSERT INTO products(name,description,category_id,price,stock_quantity,creation_date) values(\
    '{fake_product.unique.ecommerce_name()}', \
    '{fake_product.paragraph(nb_sentences=3)}', \
    {random.randint(1, NUM_OF_CATEGORIES)}, \
    {round(random.uniform(50.0, 20000.0), 2)}, \
    {random.randint(1, 500)}, \
    '{fake_product.date_between(start_date='-1y', end_date='-1d')}' \
     );")
```

```python
# генерируем данные о заказах
```

```python
NUM_OF_ORDERS = 100
```

```python
possible_order_statuses = ['Completed','Pending','Canceled']
```

```python
for i in range(NUM_OF_ORDERS):
    order_date = fake_client.date_time_between(start_date='-1y', end_date='-1m')
    print(f"INSERT INTO orders(user_id,order_date,total_amount,status,delivery_date) values(\
    {random.randint(1, NUM_OF_CLIENTS)}, \
    '{order_date}', \
    {round(random.uniform(1000.0, 100000.0), 2)}, \
    '{random.choice(possible_order_statuses)}', \
    '{order_date.date() + timedelta(days=random.randint(1, 30))}' \
     );")
```

```python
# генерируем данные о деталях заказа
```

```python
MAX_NUM_OF_POSITIONS_IN_ORDER = 5
```

```python
for i in range(NUM_OF_ORDERS):
    num_of_positions_in_order = random.randint(1, MAX_NUM_OF_POSITIONS_IN_ORDER)
    for j in range(num_of_positions_in_order):
        quantity = random.randint(1, 30)
        price_per_unit = round(random.uniform(50.0, 20000.0), 2)
        print(f"INSERT INTO order_details(order_id,product_id,quantity,price_per_unit,total_price) values(\
            {random.randint(1, NUM_OF_ORDERS)}, \
            {random.randint(1, NUM_OF_PRODUCTS)}, \
            {quantity}, \
            {price_per_unit}, \
            {round(1.0*quantity*price_per_unit, 2)} \
                 );")
```
